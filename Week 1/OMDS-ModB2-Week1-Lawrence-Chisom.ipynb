# 1. Lesson on how to search for Python commands

If you have a pandas DataFrame called df, then you can summarize its contents using df.info() or df.describe().  df.info() provides a list of column names with their counts and data types.  df.describe() will provide information such as the mean, min, max, standard deviation, and quantiles.  Thus:

```python
df = pd.DataFrame([[1, 4], [2, 5], [3, 6], [4, 7]], columns = ['A', 'B'])
df.describe()
```

In this describe() result, we see that the two columns A and B each have four elements.  The means and other statistics are shown.

```python
# Calculate each customer's age in years when the account was opened
# Divide the difference in days by 365.25 to convert it to years
df_bank_unique_accounts["Age"] = (
    (df_bank_unique_accounts["AccountOpened"] - df_bank_unique_accounts["BirthDate"]).dt.days
    / 365.25
)

# Round down to the next lower whole year
df_bank_unique_accounts["Age"] = np.floor(
    df_bank_unique_accounts["Age"]
).astype(int)

# Display the first few rows to check the new Age column
df_bank_unique_accounts[["BirthDate", "AccountOpened", "Age"]].head()
```

```python
# Find the most frequently occurring customer age
mode_age = df_bank_unique_accounts["Age"].mode()

mode_age
```

```python
# Recreate Figure 0.5 from Storytelling With Data
# This compares student feelings about science before and after a pilot program

import matplotlib.pyplot as plt
import numpy as np

# Survey response categories
categories = [
    "Bored",
    "Not great",
    "OK",
    "Kind of\ninterested",
    "Excited"
]

# Values shown in the book
before = [11, 5, 40, 25, 19]
after = [12, 6, 14, 30, 38]

# Create positions for the bars
x = np.arange(len(categories))
width = 0.35

# Create the grouped bar chart
plt.figure(figsize=(10, 6))

plt.bar(x - width/2, before, width, label="Before")
plt.bar(x + width/2, after, width, label="After")

# Add labels and title
plt.title("Pilot Program Was a Success")
plt.ylabel("Percent of Students")
plt.xlabel("How do you feel about science?")

# Add category names to the x-axis
plt.xticks(x, categories)

# Add legend
plt.legend()

plt.show()
```

I chose to recreate Figure 0.5 from the Introduction of Storytelling With Data. The graph compares how students felt about science before and after a pilot program. The biggest change is that fewer students selected “OK” after the program, while more students reported being “Kind of interested” or “Excited.” I used a grouped bar chart to make the before-and-after comparison easy to see.
